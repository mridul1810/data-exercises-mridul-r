```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('Datasets/co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

## Task 1 - A line chart showing CO2 per Capita for all countries in the dataset

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country/countries **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country/countries and its story

```python
# Task 1 
# YOUR CODE HERE
# Task 1: CO2 per capita for all countries

# Select country to highlight
highlight = 'India'

# Identify the per-capita column
per_capita_col = next(
    col for col in df.columns
    if 'capita' in col.lower()
)

# Highlight India and grey out all other countries
color_map = {
    country: 'darkblue' if country == highlight else '#EEEEEE'
    for country in df['Country'].unique()
}

# Create line chart
fig = px.line(
    df,
    x='Year',
    y=per_capita_col,
    color='Country',
    color_discrete_map=color_map,
    labels={
        'Year': '',
        per_capita_col: 'CO2 per capita'
    },
    template='plotly_white',
    height=450
)

# Remove legend and reduce visual clutter
fig.update_traces(
    line=dict(width=1),
    showlegend=False
)

# Make India's line more prominent
fig.update_traces(
    line=dict(width=3),
    selector=dict(name=highlight)
)

# Get India's first and last observations
india = df[df['Country'] == highlight].sort_values('Year')

first_value = india.iloc[0][per_capita_col]
last_value = india.iloc[-1][per_capita_col]

# Create an insight title from the data
direction = 'increased' if last_value > first_value else 'decreased'

fig.update_layout(
    title=(
        f"India's CO2 emissions per capita {direction} "
        f"from {first_value:.2f} to {last_value:.2f}"
    ),
    yaxis=dict(gridcolor='#EEEEEE'),
    xaxis=dict(showgrid=False),
    font=dict(size=13),
    margin=dict(r=120)
)

# Label India directly at the end of the line
fig.add_annotation(
    x=india.iloc[-1]['Year'],
    y=last_value,
    text='India',
    showarrow=False,
    xanchor='left',
    xshift=10,
    font=dict(
        color='darkblue',
        size=13
    )
)

fig.show()
```

## Task 2 — A line chart showing CO2 emissions over time for all Asian countries in the dataset, with one country highlighted.

**same requirements as Task1**

```python
# Task 2 
# YOUR CODE HERE
# Task — Multi-series line with highlight
# YOUR CODE HERE
# Filter Asian countries
asia = df[df['Region'] == 'Asia'].copy()

# Select country to highlight
highlight = 'China'

# Create color map
color_map = {
    country: 'crimson' if country == highlight else '#EEEEEE'
    for country in asia['Country'].unique()
}

# Create line chart
fig = px.line(
    asia,
    x='Year',
    y='CO2_Mt',
    color='Country',
    color_discrete_map=color_map,
    labels={
        'Year': '',
        'CO2_Mt': 'CO2 Emissions (Mt)'
    },
    template='plotly_white',
    height=450
)

# Thin lines and remove legend
fig.update_traces(
    line=dict(width=1),
    showlegend=False
)

# Highlight China
fig.update_traces(
    line=dict(width=3),
    selector=dict(name=highlight)
)

# Get China's emissions over time
china = asia[asia['Country'] == highlight].sort_values('Year')

first_value = china.iloc[0]['CO2_Mt']
last_value = china.iloc[-1]['CO2_Mt']

# Calculate emissions growth
growth = ((last_value - first_value) / first_value) * 100

# Add insight title
fig.update_layout(
    title=(
        f"China's CO2 emissions changed by {growth:+.0f}% "
        f"between {china.iloc[0]['Year']} and {china.iloc[-1]['Year']}"
    ),
    yaxis=dict(gridcolor='#EEEEEE'),
    xaxis=dict(showgrid=False),
    font=dict(size=13),
    margin=dict(r=120)
)

# Direct label
fig.add_annotation(
    x=china.iloc[-1]['Year'],
    y=last_value,
    text='China',
    showarrow=False,
    xanchor='left',
    xshift=10,
    font=dict(
        color='crimson',
        size=13
    )
)

fig.show()
```
